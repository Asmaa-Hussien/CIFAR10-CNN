# CIFAR-10 Image Classification using CNN
### Final CNN Lab

This notebook follows the structure of the provided **`0.0 cifar10_cnn.ipynb`** lab:
- Load CIFAR-10
- Normalize images
- Visualize samples
- Build a CNN
- Define convolution, pooling, and dense layers
- Use ReLU activation in hidden layers
- Train the model
- Plot training/validation accuracy
- Evaluate the model on the test set


```python
# 1. Import Libraries

import tensorflow as tf
import matplotlib.pyplot as plt

from tensorflow.keras import datasets, layers, models

print("TensorFlow version:", tf.__version__)
```

```python
# 2. Load CIFAR-10 Dataset

(train_images, train_labels), (test_images, test_labels) = datasets.cifar10.load_data()

print("Training images shape:", train_images.shape)
print("Training labels shape:", train_labels.shape)
print("Testing images shape:", test_images.shape)
print("Testing labels shape:", test_labels.shape)
```

## 3. Preprocessing

Each CIFAR-10 image contains RGB pixel values from **0 to 255**.

We normalize them to the range **0 to 1** by dividing by 255. This helps the neural network train more effectively.


```python
# Normalize pixel values to be between 0 and 1

train_images = train_images.astype("float32") / 255.0
test_images = test_images.astype("float32") / 255.0

print("Minimum pixel value:", train_images.min())
print("Maximum pixel value:", train_images.max())
```

```python
# 4. Class Names

class_names = [
    "airplane", "automobile", "bird", "cat", "deer",
    "dog", "frog", "horse", "ship", "truck"
]

print("Number of classes:", len(class_names))
```

```python
# 5. Display Sample Images

plt.figure(figsize=(10, 10))

for i in range(25):
    plt.subplot(5, 5, i + 1)
    plt.xticks([])
    plt.yticks([])
    plt.grid(False)
    plt.imshow(train_images[i])
    plt.xlabel(class_names[train_labels[i][0]])

plt.show()
```

# 6. Build the CNN Model

The CNN follows the same basic architecture as the provided lab.

### Layers

1. **Conv2D(32)** + ReLU
2. **MaxPooling2D**
3. **Conv2D(64)** + ReLU
4. **MaxPooling2D**
5. **Conv2D(64)** + ReLU
6. **Flatten**
7. **Dense(64)** + ReLU
8. **Dense(10)** output layer

The output contains **10 values**, one for each CIFAR-10 class.


```python
# 7. Create CNN Architecture

model = models.Sequential([
    layers.Input(shape=(32, 32, 3)),

    # Convolution Block 1
    layers.Conv2D(32, (3, 3), activation="relu"),
    layers.MaxPooling2D((2, 2)),

    # Convolution Block 2
    layers.Conv2D(64, (3, 3), activation="relu"),
    layers.MaxPooling2D((2, 2)),

    # Convolution Block 3
    layers.Conv2D(64, (3, 3), activation="relu"),

    # Classification Head
    layers.Flatten(),
    layers.Dense(64, activation="relu"),
    layers.Dense(10)
])

model.summary()
```

## 8. Activation Functions

### ReLU
ReLU is used in the convolutional and hidden dense layers:

`ReLU(x) = max(0, x)`

It introduces non-linearity and helps the network learn complex image patterns.

### Output Layer
The final layer has **10 neurons** because CIFAR-10 has 10 classes.

We keep the final layer without a softmax activation and use:

`SparseCategoricalCrossentropy(from_logits=True)`

during compilation.


```python
# 9. Compile the Model

model.compile(
    optimizer="adam",
    loss=tf.keras.losses.SparseCategoricalCrossentropy(from_logits=True),
    metrics=["accuracy"]
)

print("Model compiled successfully.")
```

## 10. Train the Model

We train for 10 epochs and use 10% of the training data for validation.


```python
# 10. Train

history = model.fit(
    train_images,
    train_labels,
    epochs=10,
    validation_split=0.1,
    batch_size=128
)
```

## 11. Training and Validation Accuracy

The following plot shows how accuracy changes during training and validation.


```python
# 11. Plot Accuracy

plt.figure(figsize=(8, 5))
plt.plot(history.history["accuracy"], label="Training Accuracy")
plt.plot(history.history["val_accuracy"], label="Validation Accuracy")
plt.xlabel("Epoch")
plt.ylabel("Accuracy")
plt.title("Training vs Validation Accuracy")
plt.legend()
plt.grid(True)
plt.show()
```

## 12. Evaluate the Model

Finally, we evaluate the trained CNN on the **unseen test dataset**.

The evaluation reports:
- Test Loss
- Test Accuracy


```python
# 12. Model Evaluation

test_loss, test_accuracy = model.evaluate(
    test_images,
    test_labels,
    verbose=2
)

print("Test Loss:", test_loss)
print("Test Accuracy:", test_accuracy)
```

```python
# 13. Final Result

print(f"Final Test Accuracy: {test_accuracy * 100:.2f}%")
```

# 14. Conclusion

The CNN successfully learns image features from the CIFAR-10 dataset.

The model uses:
- **3 convolutional layers**
- **2 max-pooling layers**
- **ReLU activation** in hidden layers
- **1 Flatten layer**
- **1 Dense hidden layer with 64 neurons**
- **10 output neurons**
- **Adam optimizer**
- **Sparse Categorical Crossentropy loss**

The final model is evaluated on the test set using **loss and accuracy**. The test accuracy printed above represents the model's performance on images that were not used during training.
